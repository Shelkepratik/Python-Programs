*remove whitespace

```python
#1)lstrip()-remove the whitespace beginning of the string
text=" pythonprogramming"
print(text.lstrip())
```

```python
#2)rstrip()-remove the whitespace end of the string
text="pythonprogramming "
print(text.rstrip())
```

```python
#3)strip()-remove the whitespace for both side
text="  pythonprogramming  "
print(text.strip())
```

*split method:

```python
#1)split():split()divides the string into multiple part and return them as list
text="python programming"
result=text.split()
print(result)
```

```python
#2)partition() :divide the string into 3 part and return the tuple
text="python is very easy language"
result=text.partition("very")
print(result)
```

```python
#2)upper() method-convert letter into uppercase
print(text.upper())
```
