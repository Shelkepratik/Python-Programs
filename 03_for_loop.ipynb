7.check prime number 

```python
num=int(input("Enter the number: "))
count=0
for i in range(1,num+1):
    if i % 2== 0:
        count+=1
if count==2:
    print("prime number")
else:
    print("not prime number")
```

9.count vowel and consonants

```python
name = input("Enter the name: ")

count = 0
consonants = 0

for i in name.lower():

    if i in "aeiou":
        count += 1
    else:
        consonants += 1

print("Count of vowels:", count)
print("Count of consonants:", consonants)
```

13.Count Even and odd number

```python
numbers=[10,15,8,21,30,17]
even=0
odd=0
for num in numbers:
    if num % 2==0:
        even +=1
    else:
        odd +=1
print("Even number: ",even)
print("Odd number: ",odd)
```

14.find maximum frequency character

```python
text=input("Enter the string: ")
max_char=""
max_count=0

for ch in text:
    count=0

    for x in text:
        if ch==x:
            count+=1

    if count > max_count:
        max_count=count
        max_char=ch
print("Maximum frequency character: ",max_char)
print("Frequency: ",max_count)
```
