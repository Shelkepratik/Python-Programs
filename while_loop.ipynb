5.Reverse each word in a sentence
Question:Write a program to reverse each word in the sentence "Hello World" using a while loop.
Expected output:olleH dlroW

```python
word="Hello world"
words=word.split()
print(words)
for char in words:
    i=len(char)-1
    while i>=0:
        print(char[i],end="")
        i-=1
    print(end=" ")
```

6.count consonants in a string
Question:Write a program to count the number of consonants in the word "learning"
input:learning
Expected output:5


```python
string="leaning"
vowel=['a','e','i','o','u']
i=0
while i <len(string):
    if string[i] not in vowel:
        print(string[i])
    i+=1
        
```
