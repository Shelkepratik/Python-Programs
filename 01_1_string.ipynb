Reverse the string using for loop

```python
str=input("Enter the string you want to reverse: ")
reverse=""
for char in str:
    reverse=char+reverse
print(reverse)
```

Reverse the String using while loop

```python
text=input("Enter the string: ")
i=len(text)-1
reverse=""
while i>=0:
    reverse=reverse+text[i]
    i-=1
print(reverse)
```

check palindrome string

```python
text=input("Enter string: ")
original=text
reverse=""
for char in text:
    reverse=char+reverse
print(reverse)
if original==reverse:
    print("Given string is palindrom")
else:
    print("Given string is not palindrom")
```

count vowel

```python
text=input("Enter the string: ")
count=0
for char in text:
    if char in "AEIOUaeiou":
        count+=1
print("vowel= ",count)
```

count vowel and consonants

```python
text=input("Enter the string: ")
vowel=0
consonants=0
for char in text:
    if char.isalpha:
        if char in "aeiouAEIOU":
            vowel+=1
        else:
            consonants+=1
print("vowel= ",vowel)
print("consonants= ",consonants)
```

find maximum frequency character

```python
text=input("Enter string: ")
max_count=0
max_char=""

for char in text:
    count=0
    for x in text:
        if char==x:
            count+=1
        

        if count > max_count:
            max_char=char
            max_count=count

       
print("maximum character in string: ",max_char)
print("maximum character count in string: ",max_count)
```

count frequency of each character

```python
text=input("Enter string; ")
frquency={}
for char in text:
    if char in frquency:
        frquency[char]=frquency[char]+1
    else:
        frquency[char]=1
print(frquency)
```

find duplicate character

```python
text=input("Enter string: ")
frquency={}
for char in text:
    if char in frquency:
        frquency[char]=frquency[char]+1
    else:
        frquency[char]=1
for char in frquency:
    if frquency[char]>1:
        print(char)
```

first Non-repeating character

```python
text = input("Enter string: ")

frequency = {}

for char in text:

    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

for char in text:

    if frequency[char] == 1:
        print("First non-repeating character =", char)
        break
```

first repeating character

```python
text=input("Enter string: ")
seen=set()
for char in text:
    if char in seen:
        print("First repeating character= ",char)
        break
    seen.add(char)
```
