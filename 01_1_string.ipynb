Reverse the string using for loop

```python
str=input("Enter the string you want to reverse: ")
reverse=""
for char in str:
    reverse=char+reverse
print(reverse)
```

Reverse the String using while loop

```python
text=input("Enter the string: ")
i=len(text)-1
reverse=""
while i>=0:
    reverse=reverse+text[i]
    i-=1
print(reverse)
```

```python
text=input("Enter string: ")
original=text
reverse=""
for char in text:
    reverse=char+reverse
print(reverse)
if original==reverse:
    print("Given string is palindrom")
else:
    print("Given string is not palindrom")
```

find maximum frequency character

```python
text=input("Enter string: ")
max_count=0
max_char=""

for char in text:
    count=0
    for x in text:
        if char==x:
            count+=1
        

        if count > max_count:
            max_char=char
            max_count=count

       
print("maximum character in string: ",max_char)
print("maximum character count in string: ",max_count)
```

first Non-repeating character

```python
text = input("Enter string: ")

frequency = {}

for char in text:

    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

for char in text:

    if frequency[char] == 1:
        print("First non-repeating character =", char)
        break
```

```python
text=input("Enter string: ")
seen=set()
for char in text:
    if char in seen:
        print("First repeating character= ",char)
        break
    seen.add(char)
```

check Anagram

```python
str1=input("Enter first string: ")
str2=input("Enter the second string: ")
if sorted(str1)==sorted(str2):
    print("Anagram")
else:
    print("not anagram")
```

anagram without sorting

```python
str1=input("Enter the first string: ")
str2=input("Enter second number: ")

if len(str1)!=len(str2):
    print("Not Anagram")
else:
    frequency1={}
    frequency2={}

    for char in str1:
        if char in frequency1:
            frequency1[char]+=1
        else:
            frequency1[char]=1
    for char in str2:
        if char in frequency2:
            frequency2[char]+=1
        else:
            frequency2[char]=1
    if frequency1==frequency2:
        print("anagram")
    else:
        print("not anagram")
```

Reverse each word

```python
text=input("Enter the string: ")
words=text.split()
result=""

for word in words:
    reverse=""
    for char in word:
        reverse=char+reverse
    result=result+reverse+" "
print(result)
```

find longest word

```python
text=input("Enter sentence: ")
words=text.split()
longest=""
for word in words:
    if len(word)>len(longest):
        longest=word
print("longest word= ",longest)
```

```python
text=input("Enter sentance; ")
words=text.split()
shortes=words[0]
for word in words:
    if len(word)<len(shortes):
        shortes=word
print("shortest word= ",shortes)
```

count alphabets,digit ans special character

```python
text=input("Enter the string: ")
alphabets=0
digit=0
special=0
for char in text:
    if char.isalpha():
        alphabets+=1
    elif char.isdigit():
        digit+=1
    else:
        special+=1
print("Alphabets= ",alphabets)
print("Digit= ",digit)
print("special character= ",special)
```
