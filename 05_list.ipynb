1.find the largest element

```python
numbers=[10,20,12,30,34,50]
largest=numbers[0]
for i in numbers:
    if i>largest:
        largest=i
print("largest element= ",largest)
```

2.find smallest element

```python
numbers=[10,20,25,40,15]
smallest=numbers[0]
for num in numbers:
    if num<smallest:
        smallest=num
print("smallest element: ",smallest)
```

3.find the sum of list

```python
numbers=[10,20,30,40]
total=0
for num in numbers:
    total=total+num
print("sum= ",total)
```

4.Reverse the list

```python
numbers=[10,20,30,40,50]
reverse=[]
for i in range(len(numbers)-1,-1,-1):
    reverse.append(numbers[i])

print(reverse)
```

5.second largest number

```python
numbers=[10,25,7,40,15]
largest=numbers[0]
second=numbers[0]

for num in numbers:
    if num>largest:
        second=largest
        largest=num
    elif num>second and num!=largest:
        second=num
print(largest)
print(second)
```

6.count even and odd number

```python
numbers=[10,15,20,23,30]
even=0
odd=0
for num in numbers:
    if num % 2==0:
        even+=1
    else:
        odd+=1
print("Even= ",even)
print("odd= ",odd)
```

7.Remove duplicate element

```python
numbers=[10,20,10,30,20,40]
result=[]

for num in numbers:
    if num not in result:
        result.append(num)
print(result)
```

8.count frequency of each element

```python
numbers=[10,20,10,30,20,10]

frequency={}

for num in numbers:
    if num in frequency:
        frequency[num]+=1
    else:
        frequency[num]=1
print(frequency)
```

9.find maximum frequency element

```python
numbers=[10,20,10,30,20,10]

frequency={}

for num in numbers:
    if num in frequency:
        frequency[num]+=1
    else:
        frequency[num]=1
print(frequency)
print(frequency[num])

max_frequency=0
max_element=None

for num in frequency:
    if frequency[num]>max_frequency:
        max_frequency=frequency[num]
        max_element=num
print("Maximum frequency element: ",max_element)
print("frequency: ",max_frequency)
```

10.Linear search

```python
numbers=[10,20,30,40,50]

search=int(input("Enter the number: "))

found=False

for i in range(len(numbers)):
    
    if numbers[i]==search:
        print("found at index ",i)
        found=True
        break
if not found:
    print("not found")
```

11.check if list is sorted

```python
numbers=[10,20,30,40,50]

sorted_list=True
for i in range(len(numbers)-1):
    if numbers[i]>numbers[i+1]:
        sorted_list=False
        break
if sorted_list:
    print("sorted")
else:
    print("not sorted")
```

12.Binary Search

```python
numbers=[10,20,30,40,50,60]

target=int(input("Enter the target"))

low=0

high=len(numbers)-1

found=False

while low<=high:

    mid=(low+high)//2

    if numbers[mid]==target:
        found=True
        break

    elif numbers[mid]<target:
        low=mid+1

    else:
        high=mid-1
if found:
    print("found")
else:
    print("Not Found")


    
```

13.move all zeros to end

```python
numbers=[0,1,0,3,12]
result=[]
zero_count=0
for i in numbers:
    if i==0:
        zero_count+=1
    else:
        result.append(i)
for i in range(zero_count):
    result.append(0)
print(result)
```

16.Rotate list by k position

```python
numbers=[1,2,3,4,5]

k=2

k=k%len(numbers)

result=numbers[-k:]+numbers[:-k]

print(result)
```

17.find missing number

```python
numbers=[1,2,3,5]

n=5

expected=n*(n+1)//2

actual=0

for num in numbers:
    actual+=num
print(actual)

missing=expected-actual

print("missing: ",missing)
```

18.find common elements

```python
list1=[10,20,30,40]
list2=[20,40,50,60]

common=[]

for num in list1:

    if num in list2:
        common.append(num)
print(common)
```

19.union of two list

```python
list1=[10,20,30]
list2=[20,30,40]

result=[]

for num in list1:
    if num not in result:
        result.append(num)

for num in list2:
    if num not in result:
        result.append(num)
print(result)
```

20.Two sum

```python
numbers=[2,7,11,15]

target=9

for i in range(len(numbers)):

    for j in range(i+1,len(numbers)):

        if numbers[i]+numbers[j]==target:
            print(i,j)
```

21.find pair with give sum

```python
numbers=[10,20,30,40,50]

target=70

for i in range(len(numbers)):

    for j in range(i+1,len(numbers)):

        if numbers[i]+numbers[j]==target:
            print(numbers[i],numbers[j])
```
