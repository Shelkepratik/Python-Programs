2.print Even number from 1 to N

```python
num=int(input("Enter the number: "))
print("Input: ",num)
for i in range(1,num+1):
    if i % 2== 0:
        print(i)
```

3.Find the sum of number from 1 to N

```python
num=int(input("Enter the number: "))
sum=0
for i in range(1,num+1):
    sum+=i
print(sum)
```

4.Print Multiplication Table

```python
num=int(input("Enter the number: "))
for i in range(1,11):
    print(num,"*",i,"=",num*i)
```

5.Count Digit in a Number

6.find the factorial of a number

```python
num=int(input("Enter the number: "))
fact=1
for i in range(1,num+1):
    fact*=i
print(fact)
```

7.check prime number 

```python
num=int(input("Enter the number: "))
count=0
for i in range(1,num+1):
    if i % 2== 0:
        count+=1
if count==2:
    print("prime number")
else:
    print("not prime number")
```

9.count vowel and consonants

```python
name = input("Enter the name: ")

count = 0
consonants = 0

for i in name.lower():

    if i in "aeiou":
        count += 1
    else:
        consonants += 1

print("Count of vowels:", count)
print("Count of consonants:", consonants)
```

10.reverse the string

```python
text=input("Enter the string")
reverse=""
for ch in text:
    reverse=ch+reverse
print("Reverse: ",reverse)
```
