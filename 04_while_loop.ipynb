5.Reverse each word in a sentence
Question:Write a program to reverse each word in the sentence "Hello World" using a while loop.
Expected output:olleH dlroW

```python
word="Hello world"
words=word.split()
print(words)
for char in words:
    i=len(char)-1
    while i>=0:
        print(char[i],end="")
        i-=1
    print(end=" ")
```

19.check palidrome number

```python
n=int(input("Enter the number; "))
original=n
reverse=0
while n>0:
    digit=n%10
    reverse=reverse*10+digit
    n//=10

if original==reverse:
    print("palindrom")
else:
    print("Not palindrom")
```

22.print prime number from 1 to N


```python
n = int(input("Enter n: "))

num = 2

while num <= n:

    i = 2
    count = 0

    while i < num:
        if num % i == 0:
            count = count + 1
        i = i + 1

    if count == 0:
        print(num)

    num = num + 1
```

24.Armstrong number

```python
n = int(input("Enter number: "))

original = n
total = 0

while n > 0:
    digit = n % 10
    total = total + digit ** 3
    n = n // 10

if total == original:
    print("Armstrong")
else:
    print("Not Armstrong")
    
```

25,Reverse the string

```python
text = input("Enter string: ")

i = len(text) - 1
reverse = ""

while i >= 0:
    reverse = reverse + text[i]
    i = i - 1

print(reverse)
```

27.count vowel and consonants

```python
text = input("Enter string: ")

i = 0
vowels = 0
consonants = 0

while i < len(text):

    if text[i].isalpha():

        if text[i] in "aeiouAEIOU":
            vowels = vowels + 1
        else:
            consonants = consonants + 1

    i = i + 1

print("Vowels =", vowels)
print("Consonants =", consonants)
```

29.find LCM of two number

```python
a=int(input("Enter the number: "))
b=int(input("Enter second number: "))
lcm=max(a,b)
while True:
    if lcm % a==0 and lcm %b==0:
        break
    lcm+=1
print("LCM= ",lcm)
```

30.check perfect number

```python
n=int(input("Enter the number: "))
i=1
total=0
while i<n:
    if n % i ==0:
        total=total+i
    i=i+1
if total ==n:
    print("perfect number")
else:
    print("not perfect number")
```
